```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('School_Data_CSV.csv')

# ---------------------------------------------------------
# a) Line plot: Number of schools per quintile across wards
# ---------------------------------------------------------
plt.figure(figsize=(12, 6))

# Group by ward_id and quintile to get counts, then unstack to create columns for each quintile
ward_quintile = df.groupby(['ward_id', 'quintile']).size().unstack(fill_value=0)

# Plot a line for each quintile
for quintile in ward_quintile.columns:
    plt.plot(ward_quintile.index, ward_quintile[quintile], marker='o', label=f'Quintile {quintile}')

plt.title('Number of Schools per Quintile Across Wards')
plt.xlabel('Ward ID')
plt.ylabel('Number of Schools')
plt.legend(title='Quintile')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# ---------------------------------------------------------
# b) Scatter plot: Geographic locations of schools
# ---------------------------------------------------------
plt.figure(figsize=(10, 8))

# Plot longitude on x-axis and latitude on y-axis
plt.scatter(df['new_long'], df['new_lat'], alpha=0.6, c='blue', edgecolors='w', label='School Locations')

plt.title('Geographic Locations of Schools')
plt.xlabel('Longitude (new_long)')
plt.ylabel('Latitude (new_lat)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# ---------------------------------------------------------
# c) Histogram: Distribution of schools by quintile
# ---------------------------------------------------------
plt.figure(figsize=(8, 6))

# Drop missing values to prevent plotting errors
valid_quintiles = df['quintile'].dropna()

plt.hist(valid_quintiles, bins=5, color='orange', edgecolor='black', alpha=0.8, label='School Count')

plt.title('Distribution of Schools by Quintile')
plt.xlabel('Quintile')
plt.ylabel('Number of Schools')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
